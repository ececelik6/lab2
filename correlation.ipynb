```python
numeric_columns = ["FSIQ", "VIQ", "PIQ", "Weight", "Height", "MRI_Count"]

for gender in ["Female", "Male"]:
    group = df[df["Gender"] == gender]

    correlation_matrix = group[numeric_columns].corr(method="pearson")

    plt.figure(figsize=(8, 6))
    sns.heatmap(
        correlation_matrix,
        annot=True,
        fmt=".2f"
    )
    plt.title(f"{gender} Pearson Correlation Heatmap")
    plt.show()
```

```python
features = ["FSIQ", "VIQ", "PIQ", "Weight", "Height"]

for gender in ["Female", "Male"]:
    group = df[df["Gender"] == gender]

    for feature in features:
        plt.figure(figsize=(6, 4))
        sns.scatterplot(
            data=group,
            x=feature,
            y="MRI_Count"
        )
        plt.title(f"{gender}: {feature} vs MRI_Count")
        plt.xlabel(feature)
        plt.ylabel("MRI_Count")
        plt.show()
```
